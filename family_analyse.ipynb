```python
import pandas as pd
import plotly.express as px
import dash
from dash import dcc, html
from dash.dependencies import Input, Output
import time

# Načtení dat
file_path = r"C:\Users\jpawlas\Desktop\ra.xlsx"
df = pd.read_excel(file_path, sheet_name='Vnoučata a pravnoučata')
df_relations = pd.read_excel(file_path, sheet_name='Děti-rodiče')

# Oprava názvů sloupců (odstranění mezer, převedení na malá písmena)
df.columns = df.columns.str.strip().str.lower()
df_relations.columns = df_relations.columns.str.strip().str.lower()

# Oprava názvu sloupce 'datum narození'
df.rename(columns={'datum narození': 'datum_narozeni'}, inplace=True)
df_relations.rename(columns={'datum narození': 'datum_narozeni'}, inplace=True)

# Převod sloupce Datum narození na datetime a vytvoření sloupce 'rok'
df['datum_narozeni'] = pd.to_datetime(df['datum_narozeni'], errors='coerce')
df_relations['datum_narozeni'] = pd.to_datetime(df_relations['datum_narozeni'], errors='coerce')
df['rok'] = df['datum_narozeni'].dt.year

df_relations['rok'] = df_relations['datum_narozeni'].dt.year

# Odstranění řádků s NaN v roce
df = df.dropna(subset=['rok'])
df_relations = df_relations.dropna(subset=['rok'])

# Konverze roku na integer
df['rok'] = df['rok'].astype(int)
df_relations['rok'] = df_relations['rok'].astype(int)

# Sloučení obou datasetů
df_combined = pd.concat([df[['jméno', 'příjmení', 'datum_narozeni', 'rok']], df_relations[['jméno', 'příjmení', 'datum_narozeni', 'rok']]])
df_combined = df_combined.sort_values('datum_narozeni').reset_index(drop=True)

# Vytvoření popisku jména a příjmení
df_combined['popisek'] = df_combined['jméno'] + ' ' + df_combined['příjmení']

# Vytvoření záchytných bodů pro jednotlivé roky
min_rok, max_rok = df_combined['rok'].min(), df_combined['rok'].max()
years_df = pd.DataFrame({'rok': list(range(min_rok, max_rok + 1)), 'datum_narozeni': [df_combined['datum_narozeni'].min()] * (max_rok - min_rok + 1)})

# Inicializace Dash aplikace
app = dash.Dash(__name__, suppress_callback_exceptions=True)

app.layout = html.Div([
    html.H1("Časová osa narození rodiny"),
    dcc.Graph(id='timeline-graph'),
    html.Button("Další", id='next-button', n_clicks=0)
], style={'backgroundColor': '#444', 'color': 'white', 'textAlign': 'center', 'padding': '20px'})

@app.callback(
    Output('timeline-graph', 'figure'),
    Input('next-button', 'n_clicks')
)
def update_graph(n_clicks):
    if n_clicks == 0:
        return px.scatter()
    
    data_to_show = df_combined.iloc[:n_clicks]
    
    fig = px.scatter(data_to_show, x='datum_narozeni', y='rok',
                     text='popisek', title="Časová osa narození",
                     template='plotly_dark')
    
    # Přidání záchytných bodů pro roky
    for year in years_df['rok']:
        fig.add_shape(
            type="line",
            x0=years_df['datum_narozeni'].min(), x1=years_df['datum_narozeni'].max(),
            y0=year, y1=year,
            line=dict(color="gray", width=0.5, dash="dot")
        )
    
    fig.update_layout(
    yaxis=dict(
        title="Rok narození",
        tickmode="array",
        tickvals=list(range(min_rok, max_rok + 1)),  # Hodnoty na ose Y
        ticktext=[str(year) for year in range(min_rok, max_rok + 1)]  # Popisky k hodnotám
    ),
        xaxis=dict(title="Datum narození"),
        plot_bgcolor='#555',
        paper_bgcolor='#555'
    )
    return fig

if __name__ == '__main__':
    app.run_server(debug=True, host='127.0.0.1', port=8050)
```

```python
import pandas as pd
import plotly.express as px
import dash
from dash import dcc, html
from dash.dependencies import Input, Output

# Načtení dat
file_path = r"C:\Users\jpawlas\Desktop\ra.xlsx"
df_relations = pd.read_excel(file_path, sheet_name='Děti-rodiče')

# Oprava názvů sloupců
df_relations.columns = df_relations.columns.str.strip().str.lower()
df_relations.rename(columns={'datum narození': 'datum_narozeni', 'pohlavi': 'pohlavi'}, inplace=True)

# Převod datového typu
df_relations['datum_narozeni'] = pd.to_datetime(df_relations['datum_narozeni'], errors='coerce')

# Přidání sloupce s celým jménem
df_relations['full_name'] = df_relations['jméno'] + ' ' + df_relations['příjmení']

# Vytvoření souřadnic Y podle pohlaví
male_offset = 0.5
female_offset = -0.5
step = 0.3

# Použití _asdict pro správný přístup k hodnotám sloupců
df_relations['y_pos'] = [
    (i * step + male_offset) if row.pohlavi == 'm' else (i * step + female_offset)
    for i, row in enumerate(df_relations.itertuples())
]

# Seřazení dat podle data narození (chronologicky)
df_relations = df_relations.sort_values(by='datum_narozeni')

# Inicializace Dash aplikace
app = dash.Dash(__name__, suppress_callback_exceptions=True)

# Layout s tlačítky
app.layout = html.Div([
    html.H1("Časová osa narození - Děti a rodiče"),
    dcc.Graph(id='timeline-graph'),
    html.Button("Další", id='next-button', n_clicks=0),
    html.Button("Zpět", id='prev-button', n_clicks=0)
], style={'backgroundColor': '#222', 'color': 'white', 'textAlign': 'center', 'padding': '20px'})

@app.callback(
    Output('timeline-graph', 'figure'),
    [Input('next-button', 'n_clicks'),
     Input('prev-button', 'n_clicks')]
)
def update_graph(n_clicks_next, n_clicks_prev):
    # Počet zobrazených bodů (dynamické)
    n_to_show = n_clicks_next - n_clicks_prev
    
    # Pokud není žádný bod k zobrazení, vrátíme prázdný scatter
    if n_to_show <= 0:
        return px.scatter()

    data_to_show = df_relations.iloc[:n_to_show]
    
    fig = px.scatter(data_to_show, x='datum_narozeni', y='y_pos',
                     text='full_name', title="Časová osa narození",
                     template='plotly_dark')
    
    fig.update_traces(textposition='top center', marker=dict(size=10))
    
    fig.update_layout(
        yaxis=dict(showticklabels=False, title=""),
        xaxis=dict(title="Datum narození", dtick="M12"),
        plot_bgcolor='#222',
        paper_bgcolor='#222'
    )
    
    return fig

if __name__ == '__main__':
    app.run_server(debug=True, host='127.0.0.1', port=8051)

```
