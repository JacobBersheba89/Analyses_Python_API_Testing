```python
import pandas as pd
import plotly.express as px
import dash
from dash import dcc, html
from dash.dependencies import Input, Output
import time

# Načtení dat
file_path = r"C:\Users\jpawlas\Desktop\ra.xlsx"
df = pd.read_excel(file_path, sheet_name='Vnoučata a pravnoučata')

# Převod sloupce Datum narození na datetime a seřazení podle něj
df['Datum narození'] = pd.to_datetime(df['Datum narození'], errors='coerce')
df = df.sort_values('Datum narození').reset_index(drop=True)

# Inicializace Dash aplikace
app = dash.Dash(__name__)

app.layout = html.Div([
    html.H1("Časová osa narození rodiny"),
    dcc.Graph(id='timeline-graph'),
    html.Button("Další", id='next-button', n_clicks=0)
])

@app.callback(
    Output('timeline-graph', 'figure'),
    Input('next-button', 'n_clicks')
)
def update_graph(n_clicks):
    if n_clicks == 0:
        return px.scatter()
    
    data_to_show = df.iloc[:n_clicks]
    
    fig = px.scatter(data_to_show, x='Datum narození', y=['Jméno'],
                     text='Jméno', title="Časová osa narození")
    fig.update_traces(textposition='top center')
    return fig

if __name__ == '__main__':
    app.run_server(debug=True)
```
